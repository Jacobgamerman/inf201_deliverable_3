```python
#Task 5 Running battles
from battle import Battle
def run_battles(hero_name: str, enemy_name: str, weapon_name: str, class_1: type, class_2: type) -> None:
    """
    Run simulations across all weapon combinations for any two character classes:
      1. Both unarmed
      2. Hero armed, enemy unarmed
      3. Hero unarmed, enemy armed
      4. Both armed
    """
    weapon_configs = [
        ("Both Unarmed", False, False),
        (f"{hero_name} Armed, {enemy_name} Unarmed", True, False),
        (f"{hero_name} Unarmed, {enemy_name} Armed", False, True),
        ("Both Armed", True, True),
    ]

    for label, hero_armed, enemy_armed in weapon_configs:
        print(f"\n{'=' * 50}")
        print(f"Setup: {class_1.__name__} vs {class_2.__name__} | {label}")
        print(f"{'=' * 50}")

        # Instantiate fresh weapons with a standard d8/d12 die
        hero_weapon = Weapon(name=weapon_name, damage_dice=Dice(sides=8)) if hero_armed else None
        enemy_weapon = Weapon(name=weapon_name, damage_dice=Dice(sides=8)) if enemy_armed else None

        # Fresh character instances each round so HP/mana reset
        hero = class_1(name=hero_name, weapon=hero_weapon)
        enemy = class_2(name=enemy_name, weapon=enemy_weapon)

        print(f"Hero:  {hero}")
        print(f"Enemy: {enemy}\n")

        battle = Battle(hero=hero, enemy=enemy)
        battle.simulate_battle()


def run_all_class_combinations():
    """Run all class matchups across the simulator."""
    classes = [Character, Warrior, Mage]
    
    # Test every pair combination
    for i in range(len(classes)):
        for j in range(i, len(classes)):
            cls1, cls2 = classes[i], classes[j]
            run_battles(
                hero_name="Carl",
                enemy_name="Donut",
                weapon_name="Flail",
                class_1=cls1,
                class_2=cls2
            )

# Execute in your notebook cell:
run_all_class_combinations()
```

Test of different battles. Here we see a bias toward the character starting; the hero or here Carl. This could be balanced with a dice roll deciding the starting player.
With an armor of 5 and a d8 the chance of damaging the enemy is quite low. Without a lower unarmed damage or higher base damage, where the dice is an extra being unarmed is a better move
